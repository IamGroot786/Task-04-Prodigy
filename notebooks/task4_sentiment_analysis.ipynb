# Prodigy Infotech Task 4 — Twitter Entity Sentiment Analysis

This notebook uses the official Task 4 `twitter_training.csv` sample. Run `python src/fetch_data.py` first. The loader also includes `twitter_validation.csv` if you obtain and place it in `data/raw/`, but the official folder path checked for this project did not expose that file. The four source labels—Positive, Neutral, Negative, and Irrelevant—are retained.


```python
from pathlib import Path
import sys, json
import pandas as pd
from IPython.display import display, Image
ROOT = Path.cwd()
if not (ROOT / 'src').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from analyze_sentiment import load_source_files, clean_dataframe, make_outputs, FIGURES, REPORTS
raw = load_source_files()
print('Raw combined shape:', raw.shape)
display(raw.head())
```

## Data quality and cleaning audit
Missing/empty text, invalid sentiment labels, and duplicate records are audited before plotting.

```python
cleaned, audit = clean_dataframe(raw)
print(json.dumps(audit, indent=2, ensure_ascii=False))
display(cleaned.head())
print(f'Retained {len(cleaned):,} of {len(raw):,} source rows')
```

## Sentiment and entity analysis
The outputs include overall sentiment distribution, most-discussed entities, sentiment mix by entity, positive-minus-negative share, and frequent terms per sentiment. A training-versus-validation comparison is produced only if both source files are available. Interpret proportions alongside sample counts.


```python
REPORTS.mkdir(parents=True, exist_ok=True)
(REPORTS / 'cleaning_audit.json').write_text(json.dumps(audit, indent=2), encoding='utf-8')
report = make_outputs(cleaned, audit)
display(pd.DataFrame({'count': report['sentiment_counts'], 'percentage': report['sentiment_percentages']}))
print('Executive summary:', REPORTS / 'executive_summary.md')
```

```python
for filename in ['sentiment_distribution.png','sentiment_by_source_split.png','top_entities_by_volume.png','sentiment_by_entity.png','positive_negative_balance_by_entity.png','top_terms_positive.png','top_terms_negative.png','top_terms_neutral.png','top_terms_irrelevant.png']:
    path = FIGURES / filename
    if path.exists():
        print(filename)
        display(Image(filename=str(path)))
```

## Interpretation checklist
- Report observed counts and percentages from the generated outputs only.
- The source sentiment labels are inherited and not independently verified.
- Irrelevant is distinct from Neutral.
